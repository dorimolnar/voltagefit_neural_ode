This notebook generates the figures used in my thesis about the experimetns with Pospischil-type channels.

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
```

```python
base_trained_preds_25 = np.load("../results/base_trained_preds.npy")
base_trained_losses_25 = np.load("../results/base_trained_losses.npy")
catkm_trained_preds_25 = np.load("../results/catkm_trained_preds.npy")
catkm_trained_losses_25 = np.load("../results/catkm_trained_losses.npy")
base_trained_nn_preds_25 = np.load("../results/base_trained_nn_preds.npy")
base_trained_nn_losses_25 = np.load("../results/base_trained_nn_losses.npy")
catkm_trained_nn_preds_25 = np.load("../results/catkm_trained_nn_preds.npy")
catkm_trained_nn_losses_25 = np.load("../results/catkm_trained_nn_losses.npy")

v_data_25 = np.load("../results/v_data.npy")
base_init_25 = np.load("../results/base_init.npy")
base_init_losses_25 = np.load("../results/base_init_losses.npy")
catkm_init_25 = np.load("../results/catkm_init.npy")
catkm_init_losses_25 = np.load("../results/catkm_init_losses.npy")

time_vec_25 = np.arange(0, 150+2*0.025, 0.025)
time_vec_25 = time_vec_25[1500:]
```

```python
with mpl.rc_context(fname="../../../.matplotlibrc"):
    fig, axs = plt.subplots(1,5, figsize=(10, 1.5))
    fig.subplots_adjust(wspace=0.4)

    for i in range(5):
        axs[i].plot(time_vec_25, v_data_25[0], label="Experimental target", linewidth=1.5)
        axs[i].plot(time_vec_25, base_init_25[i], c='C1', label="Single-compartment prediction", linestyle="--", linewidth=1.5)
        axs[i].set_title(f"Loss: {base_init_losses_25[i]:.4f}")
        axs[i].set_ylim(-100, 30)

    axs[0].set_ylabel("Voltage (mV)")

    fig.savefig("../svg/base_init_25.svg", bbox_inches='tight', dpi=300)

    plt.tight_layout()
    plt.show()
```

```python
with mpl.rc_context(fname="../../../.matplotlibrc"):
    fig, axs = plt.subplots(1,5, figsize=(10, 1.5))
    fig.subplots_adjust(wspace=0.4)

    for i in range(5):
        axs[i].plot(time_vec_25, v_data_25[0], label="Experimental target", linewidth=1.5)
        axs[i].plot(time_vec_25, catkm_init_25[i], c='C3', label="Single-compartment prediction", linestyle="--", linewidth=1.5)
        axs[i].set_title(f"Loss: {catkm_init_losses_25[i]:.4f}")
        axs[i].set_ylim(-100, 30)
        axs[i].set_xlabel("Time (ms)")

    axs[0].set_ylabel("Voltage (mV)")

    fig.savefig("../svg/catkm_init_25.svg", bbox_inches='tight', dpi=300)

    plt.tight_layout()
    plt.show()
```

```python
from jaxley.solver_gate import (
    save_exp
)
import jax.numpy as jnp

def linear_interp(x, xp, fp):
    """Linear interpolation for JAX."""
    idx = jnp.clip(jnp.searchsorted(xp, x, side="right") - 1, 0, len(xp) - 2)
    x0 = xp[idx]
    x1 = xp[idx + 1]
    y0 = fp[idx]
    y1 = fp[idx + 1]
    slope = (y1 - y0) / (x1 - x0 + 1e-8)
    return y0 + slope * (x - x0)


def interpolate_params(v, x_vals, params):
    interp = linear_interp(v, x_vals, params)
    return interp

def efun(x):
    """x/[exp(x)-1]

    Args:
        x (float): _description_

    Returns:
        float: x/[exp(x)-1]
    """
    return x / (save_exp(x) - 1.0)

def m_gate(v, vt, x_vals = None, m_alpha_params = None, m_beta_params = None):
    if x_vals is None:
        x_vals = jnp.linspace(-100, 50, 10)
    v_alpha = v - vt - 13.0
    alpha = 0.32 * efun(-0.25 * v_alpha) / 0.25

    v_beta = v - vt - 40.0
    beta = 0.28 * efun(0.2 * v_beta) / 0.2
    alpha = jnp.abs(alpha + interpolate_params(v, x_vals, m_alpha_params))
    beta = jnp.abs(beta + interpolate_params(v, x_vals, m_beta_params))
    return alpha, beta

def h_gate(v, vt, x_vals = None, h_alpha_params = None, h_beta_params = None):
    if x_vals is None:
        x_vals = jnp.linspace(-100, 50, 10)
    v_alpha = v - vt - 17.0
    alpha = 0.128 * save_exp(-v_alpha / 18.0)

    v_beta = v - vt - 40.0
    beta = 4.0 / (save_exp(-v_beta / 5.0) + 1.0)

    alpha = jnp.abs(alpha + interpolate_params(v, x_vals, h_alpha_params))
    beta = jnp.abs(beta + interpolate_params(v, x_vals, h_beta_params))
    return alpha, beta

def n_gate(v, vt, x_vals = None, n_alpha_params = None, n_beta_params = None):
    if x_vals is None:
        x_vals = jnp.linspace(-100, 50, 10)
    v_alpha = v - vt - 15.0
    alpha = 0.032 * efun(-0.2 * v_alpha) / 0.2

    v_beta = v - vt - 10.0
    beta = 0.5 * save_exp(-v_beta / 40.0)

    alpha = jnp.abs(alpha + interpolate_params(v, x_vals, n_alpha_params))
    beta = jnp.abs(beta + interpolate_params(v, x_vals, n_beta_params))
    return alpha, beta

def u_gate(v, vx, x_vals = None, u_inf_params = None, u_tau_params = None):
    if x_vals is None:
        x_vals = jnp.linspace(-100, 50, 10)
    v_u1 = v + vx + 81.0
    u_inf = 1.0 / (1.0 + save_exp(v_u1 / 4))

    tau_u = (30.8 + (211.4 + save_exp((v + vx + 113.2) / 5.0))) / (
        3.7 * (1 + save_exp((v + vx + 84.0) / 3.2))
    )

    u_inf = jnp.abs(u_inf + interpolate_params(v, x_vals, u_inf_params))
    tau_u = jnp.abs(tau_u + interpolate_params(v, x_vals, u_tau_params))

    return u_inf, tau_u


voltages = np.linspace(-100, 50, 200)
am, bm = m_gate(voltages, -60, None, inv_Na_gating_params[:10], inv_Na_gating_params[10:20])
am_0, bm_0 = m_gate(voltages, -60, None, jnp.zeros(10), jnp.zeros(10))
ah, bh = h_gate(voltages, -60, None, inv_Na_gating_params[20:30], inv_Na_gating_params[30:40])
ah_0, bh_0 = h_gate(voltages, -60, None, jnp.zeros(10), jnp.zeros(10))
an, bn = n_gate(voltages, -60, None, inv_K_gating_params[:10], inv_K_gating_params[10:20])
an_0, bn_0 = n_gate(voltages, -60, None, jnp.zeros(10), jnp.zeros(10))
u_inf, u_tau = u_gate(voltages, 2.0, None, inv_Ca_gating_params[:10], inv_Ca_gating_params[10:20])
u_inf_0, u_tau_0 = u_gate(voltages, 2.0, None, jnp.zeros(10), jnp.zeros(10))

with mpl.rc_context(fname="../../../.matplotlibrc"):
    fig, axs = plt.subplots(2,4, figsize = (10,4))
    fig.subplots_adjust(hspace=0.2, wspace=0.3)

    axs[0,0].plot(voltages, am_0, c = "violet", linewidth=1.2)
    axs[0,0].plot(voltages, am, c = "deepskyblue", linewidth=1.0)
    axs[0,0].set_title('m_gate')
    axs[0,0].set_ylabel('alpha')
    axs[1,0].plot(voltages, bm_0, c = "violet", linewidth=1.2)
    axs[1,0].plot(voltages, bm, c = "deepskyblue", linewidth=1.0)
    axs[1,0].set_ylabel('beta')
    axs[1,0].set_xlabel('Voltage (mV)')
    axs[0,1].plot(voltages, ah_0, c = "violet", linewidth=1.2)
    axs[0,1].plot(voltages, ah, c = "deepskyblue", linewidth=1.0)
    axs[0,1].set_title('h_gate')
    axs[1,1].plot(voltages, bh_0, c = "violet", linewidth=1.2)
    axs[1,1].plot(voltages, bh, c = "deepskyblue", linewidth=1.0)
    axs[1,1].set_xlabel('Voltage (mV)')
    axs[0,2].plot(voltages, an_0, c = "violet", linewidth=1.2, label = "Original function")
    axs[0,2].plot(voltages, an, c = "deepskyblue", linewidth=1.0, label = "Learned function")
    axs[0,2].set_title('n_gate')
    axs[1,2].plot(voltages, bn_0, c = "violet", linewidth=1.2)
    axs[1,2].plot(voltages, bn, c = "deepskyblue", linewidth=1.0)
    axs[1,2].set_yticks([0.0, 0.5, 1.0, 1.5])
    axs[1,2].set_xlabel('Voltage (mV)')
    axs[0,3].plot(voltages, u_inf_0, c = "violet", linewidth=1.2, label = "Standard function")
    axs[0,3].plot(voltages, u_inf, c = "deepskyblue", linewidth=1.0, label = "Learned function")
    axs[0,3].set_title('u_gate')
    axs[0,3].set_ylabel('u_inf')
    axs[0,3].legend()
    axs[1,3].plot(voltages, u_tau_0, c = "violet", linewidth=1.2)
    axs[1,3].plot(voltages, u_tau, c = "deepskyblue", linewidth=1.0)
    axs[1,3].set_ylabel('tau_u')
    axs[1,3].set_xlabel('Voltage (mV)')

    fig.savefig("../svg/gating_params_catkm_trained_25.svg", bbox_inches='tight', dpi=300)

    plt.show()
```
